```
def fibonacci(n):
    # Se 'n' for menor ou igual a 0, não há elementos para retornar
    if n <= 0:
        return []

    # Se 'n' for 1, retorna apenas o primeiro número da sequência
    elif n == 1:
        return [0]

    # Para 'n' maior ou igual a 2, iniciamos a lista com os dois primeiros números
    sequencia = [0, 1]

    # O loop continua adicionando números enquanto o tamanho da lista for menor que 'n'
    while len(sequencia) < n:
        # sequencia[-1] pega o último número e sequencia[-2] pega o penúltimo
        proximo = sequencia[-1] + sequencia[-2]

        # Adiciona o novo número calculado ao final da lista
        sequencia.append(proximo)

    # Devolve a lista pronta com os 'n' números
    return sequencia


# Exemplo de teste: gera os primeiros 8 números
resultado = fibonacci(8)
print(resultado)  # Saída: [0, 1, 1, 2, 3, 5, 8, 13]
```

```
#def: avisa que to criando definitivo uma função
def eh_palindromo(texto):
    #minusculo texto
    texto = texto.lower()
    #sem espaço em branco
    texto = texto.replace(" ", "")
    #compara texto sem nada com ele mesmo, mas invertido
    return texto == texto[::-1]

palavra = ["arara", "python", "A base do teto desaba"]

for p in palavra:
    resultado = eh_palindromo(p)
    print(f"{p}: {resultado}")
```

```
def bubble_sort(lista):
    #len(length):medir o tamanho ou contar a quantidade de itens em uma estrutura de dados
    n = len(lista)

    # Percorre toda a lista
    for i in range(n):
        # Variável para verificar se houve alguma troca nesta passada
        trocou = False

        # Últimos 'i' elementos já estão no lugar certo, não precisamos olhar para eles
        for j in range(0, n - i - 1):

            # Compara o elemento atual com o próximo vizinho
            if lista[j] > lista[j + 1]:
                # Troca os elementos de posição em Python (atribuição múltipla)
                lista[j], lista[j + 1] = lista[j + 1], lista[j]

                # Marca que houve pelo menos uma troca nesta rodada
                trocou = True

        # Se nenhuma troca ocorreu, a lista já está perfeitamente ordenada!
        if not trocou:
            break

    return lista

numeros = [5,18,2,63,109,1]
print("Lista original:", numeros)

ordenados = bubble_sort(numeros)
print("Lista ordenada:", ordenados)
```
